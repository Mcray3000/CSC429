```python
import kagglehub
import shutil
import os

path = kagglehub.dataset_download("hojjatk/mnist-dataset")

destination_dir = "./mnist_data"

shutil.copytree(path, destination_dir, dirs_exist_ok=True)

print(f"Dataset successfully copied to: {os.path.abspath(destination_dir)}")
```
